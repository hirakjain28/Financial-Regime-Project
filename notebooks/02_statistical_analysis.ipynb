SECTION 1 — Load Clean Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.seasonal import seasonal_decompose

nifty = pd.read_csv("../data/processed/nifty_clean.csv", index_col='Date', parse_dates=True)

nifty.head()
```

SECTION 2 — STATIONARITY

```python
result = adfuller(nifty['Close'].dropna())

print("ADF Statistic:", result[0])
print("p-value:", result[1])
```

SECTION 3 — DIFFERENCING

```python
nifty['Diff_Close'] = nifty['Close'].diff()

plt.plot(nifty['Diff_Close'])
plt.title("Differenced Close Price")
plt.show()
```

SECTION 4 — ACF

```python
plot_acf(nifty['Returns'].dropna(), lags=40)
plt.show()
```

SECTION 5 — PACF

```python
plot_pacf(nifty['Returns'].dropna(), lags=40)
plt.show()
```

SECTION 8 — VOLATILITY CLUSTERING

```python
plt.plot(nifty['Volatility'])
plt.title("Volatility Clustering")
plt.show()
```
